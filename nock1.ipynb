```python
#Pandasインポートとデータ読み込み
import pandas as pd

customer_master = pd.read_csv("../Data/SampleCode/Chapter1/customer_master.csv")
customer_master.head()
```

```python
item_master = pd.read_csv("../Data/SampleCode/Chapter1/item_master.csv")
item_master.head()
```

```python
transaction_1 = pd.read_csv("../Data/SampleCode/Chapter1/transaction_1.csv")
transaction_2 = pd.read_csv("../Data/SampleCode/Chapter1/transaction_2.csv")
transaction_1.head()
```

```python
transaction_detail_1 = pd.read_csv("../Data/SampleCode/Chapter1/transaction_detail_1.csv")
transaction_detail_1.head()
```

```python
#取引データを結合(ユニオン)：行方向結合する
transaction = pd.concat([transaction_1,transaction_2],ignore_index=True)
transaction.head()
```

```python
#正しく結合できてるか確認
print(len(transaction_1)) 
print(len(transaction_2))
print(len(transaction))
```

```python
#より粒度の細かい取引データのほうも同様にまとめる
transaction_detail_2 = pd.read_csv("../Data/SampleCode/Chapter1/transaction_detail_2.csv")
transaction_detail = pd.concat([transaction_detail_1,transaction_detail_2],ignore_index=True)
transaction_detail.head()
```

```python
#結合(ジョイン)：列方向結合する
#粒度の異なるデータなのでお互いに足りないデータを補完
join_data = pd.merge(transaction_detail, transaction[["transaction_id","payment_date","customer_id"]],
                     on="transaction_id", how="left")
join_data.head()
```

```python
#正しく結合できてるか確認
print(len(transaction_detail))
print(len(transaction))
print(len(join_data))
```

```python
#データ項目を追加する
join_data = pd.merge(join_data,customer_master,on="customer_id",how="left")
join_data = pd.merge(join_data,item_master,on="item_id",how="left")
join_data.head()
```

```python
#必要なデータ列を作る
join_data["price"] = join_data["quantity"]*join_data["item_price"]
join_data[["quantity","item_price","price"]].head()
```

```python
#データの検算
#case1：合計値で確認
print(join_data["price"].sum())
print(transaction["price"].sum())

#case2：true falseで確認
join_data["price"].sum()==transaction["price"].sum()
```

```python
#欠損値の確認
join_data.isnull().sum()
```

```python
#全体の数字感（統計量）の確認
join_data.describe()
```

```python
#データの期間範囲の確認
print(join_data["payment_date"].min())
print(join_data["payment_date"].max())
```

```python
#データの型の確認
join_data.dtypes
```

```python
#時間を扱うのでobject型からdatetime型に変換する
join_data["payment_date"]=pd.to_datetime(join_data["payment_date"])
join_data["payment_month"]=join_data["payment_date"].dt.strftime("%Y%m")
join_data[["payment_date","payment_month"]].head()
```

```python
#データを集計するgroupby(集計する列)[表示するもの].集計方法
join_data.groupby("payment_month")["price"].sum()
```

```python
#月別,商品別にデータを集計する
join_data.groupby(["payment_month","item_name"])[["price","quantity"]].sum()
```

```python
#醜いのでvivot_tableで見やすくする
pd.pivot_table(join_data,index="item_name",columns="payment_month",values=["price","quantity"], aggfunc="sum")
```

```python
#データを可視化するためのグラフ用データの作成
graph_data=pd.pivot_table(join_data,index="payment_month",columns="item_name",values="price",aggfunc="sum")
graph_data.head()
```

```python
#グラフを描画する
import matplotlib.pyplot as plt

plt.plot(list(graph_data.index),graph_data["PC-A"],label="PC-A")
plt.plot(list(graph_data.index),graph_data["PC-B"],label="PC-B")
plt.plot(list(graph_data.index),graph_data["PC-C"],label="PC-C")
plt.plot(list(graph_data.index),graph_data["PC-D"],label="PC-D")
plt.plot(list(graph_data.index),graph_data["PC-E"],label="PC-E")
plt.title("Sales trends")
plt.legend()
```
